### 9. Dataset Sementara

```python
dataset_copy = dataset.copy()
```

#### *b. Data Preprocessing*

##### 1. Encode Target

```python
dataset_copy['Target'] = dataset_copy['Delivery_Time_min'].apply(lambda x: 1 if x < 41 else (2 if x > 71 else 3))
```

```python
dataset_copy['Target'].value_counts()
```

Inspeksi distribusi kelas target yang ada di dataset

```python
target_count = dataset_copy['Target'].value_counts()
explode = [0.05] * len(target_count)

plt.figure(figsize=(10,10))
plt.pie(
    target_count,
    explode=explode,
    labels=target_count.index.map({1: "Lama", 2: "Lumayan", 3: "Sebentar"}),
    autopct='%1.1f%%',
    startangle=140
)
plt.show()
```
