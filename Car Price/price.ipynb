```python
def data_range(dataset, col):
    for cols in col:
        Q1 = dataset[cols].quantile(0.25)
        Q3 = dataset[cols].quantile(0.75)

        IQR = Q3 - Q1

        print(f"Rentang Data Kolom: {cols}")
        print(f"Q1: {Q1}")
        print(f"Q3: {Q3}")
        print(f"IQR: {IQR}\n")

data_range(dataset, fitur_continu)
```

```python
data_range(dataset, ['enginesize', 'compressionratio', 'horsepower'])
```
